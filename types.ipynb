```python
from typing import TypedDict

class Movie(TypedDict):
    title: str
    year: int

movie = Movie(title="The Dark Knight", year=2008)

print(movie)
```

```python
from typing import Union

def Square(x: Union[int, float]) -> float:
    return x * x

try:
    print(Square(2))
    print(Square(2.5))    
    print(Square("invalid"))
except TypeError as e:
    print(f"Error: {e}")
```

```python
from typing import Optional

def greet(name: Optional[str] = None) -> str:
    if name:
        return f"Hello, {name}!"
    else:
        return "Hello, World!"
    
print(greet())
print(greet("Alice"))
```

```python
cubed = lambda x: x ** 3
print(cubed(3))

nums = [1, 2, 3, 4, 5]
cubes = list(map(cubed, nums))
print(cubes)

def print_value(x: int) -> str:
    return f"The value is {x}"

print(print_value(10))
print()

print_values = list(map(print_value, nums))
print(*print_values, sep="\n") # unpacking operator, ** is for dictionaries
print("====")

# alternate with worse performance (because of intermediate objects)
list(map(lambda x: print(print_value(x)), nums))
```
